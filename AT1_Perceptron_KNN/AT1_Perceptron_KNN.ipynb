```python
# importando a biblioteca NumPy para manipulação de arrays e matrizes
import numpy as np

# Matriz de treino: [Valor Normalizado (0.0 a 10.0), Frequência de Operações]
X_treino_fraude = np.array([
    [1.5, 1.0],
    [2.0, 2.0],
    [3.5, 1.5],
    [3.0, 3.0],
    [6.5, 5.0],
    [7.0, 7.0],
    [8.5, 6.0],
    [9.0, 8.0]
])

# Rótulos das classes (0: Legítima, 1: Suspeita)
y_treino_fraude = np.array([0, 0, 0, 0, 1, 1, 1, 1])
```

```python
# Função para treinar o perceptron
def treinar_perceptron(X, y, taxa_aprendizado, epocas):
    w = np.zeros(X.shape[1])
    b = 0 
    
# repetindo o processo de treinamento por um número definido de épocas
    for _ in range(epocas):
        for i in range(X.shape[0]):
            y_pred = np.dot(X[i], w) + b
            erro = y[i] - (1 if y_pred >= 0 else 0)
            w += taxa_aprendizado * erro * X[i]
            b += taxa_aprendizado * erro

            return w, b
```

```python
# Função para prever a classe de uma nova transação
w, b = treinar_perceptron(
    X_treino_fraude,
    y_treino_fraude,
    taxa_aprendizado=0.1,
    epocas=10
)

print("Pesos aprendidos:", w)
print("Viés aprendido:", b)
```

```python
#
def prever_perceptron(X, w, b):
    z = np.dot(X, w) + b
    return 1 if z >= 0 else 0
```

```python
transacao_A = np.array([5.0, 4.0])
transacao_B = np.array([7.0, 6.5])
```

```python
previsao_A = prever_perceptron(transacao_A, w, b)
previsao_B = prever_perceptron(transacao_B, w, b)

print("Transação A:", "Suspeita" if previsao_A == 1 else "Legítima")
print("Transação B:", "Suspeita" if previsao_B == 1 else "Legítima")
```

```python
# Predição de Risco de Churn com Classificador KNN

X_treino_churn = np.array([
    [2.0, 1.0],
    [3.0, 0.0],
    [5.0, 1.0],
    [6.0, 2.0],
    [15.0, 4.0],
    [18.0, 5.0],
    [20.0, 4.0],
    [22.0, 6.0]
])

y_treino_churn = np.array([
    0, 0, 0, 0,
    1, 1, 1, 1
])
```

```python
def classificar_knn(X_treino, y_treino, ponto, k, metrica):
    diferencas = X_treino - ponto

    if metrica == "euclidiana":
        distancias = np.sqrt(np.sum(diferencas ** 2, axis=1))

    elif metrica == "manhattan":
        distancias = np.sum(np.abs(diferencas), axis=1)

    else:
        raise ValueError("Métrica deve ser 'euclidiana' ou 'manhattan'")

    indices = np.argsort(distancias)[:k]

    classes_vizinhos = y_treino[indices]

    valores, contagens = np.unique(classes_vizinhos, return_counts=True)

    classe_predita = valores[np.argmax(contagens)]

    return classe_predita, indices, distancias[indices]
```

```python
cliente_1 = np.array([4.0, 1.0])
cliente_2 = np.array([17.0, 5.0])

resultado_1 = classificar_knn(
    X_treino_churn,
    y_treino_churn,
    cliente_1,
    k=3,
    metrica="euclidiana"
)

resultado_2 = classificar_knn(
    X_treino_churn,
    y_treino_churn,
    cliente_2,
    k=3,
    metrica="euclidiana"
)
```

```python
classe_1, indices_1, distancias_1 = resultado_1
classe_2, indices_2, distancias_2 = resultado_2

print("Cliente 1")
print("Classe:", "Alto Risco" if classe_1 == 1 else "Baixo Risco")
print("Índices dos vizinhos:", indices_1)
print("Distâncias:", distancias_1)

print("\nCliente 2")
print("Classe:", "Alto Risco" if classe_2 == 1 else "Baixo Risco")
print("Índices dos vizinhos:", indices_2)
print("Distâncias:", distancias_2)
```

```python
#Recomendação de Servidores Cloud por Similaridade Espacial
catalogo_servidores = np.array([
    [2.0, 4.0, 50.0],
    [4.0, 8.0, 100.0],
    [8.0, 16.0, 250.0],
    [16.0, 32.0, 500.0],
    [32.0, 64.0, 1000.0],
    [64.0, 128.0, 2000.0]
])

nomes_servidores = [
    "Micro Instância Web",
    "Standard App Server",
    "Medium Backend & Cache",
    "Database Enterprise",
    "High Performance Computing",
    "Big Data & AI Training"
]

perfil_demanda = np.array([12.0, 28.0, 850.0])

def recomendar_servidores(catalogo, demanda, k):
    diferencas = catalogo - demanda

    distancias = np.sqrt(
        np.sum(diferencas ** 2, axis=1)
    )

    indices = np.argsort(distancias)[:k]

    return indices, distancias[indices]

indices_recomendados, distancias_recomendadas = recomendar_servidores(
    catalogo_servidores,
    perfil_demanda,
    k=2
)

print("Índices recomendados:", indices_recomendados)
print("Distâncias:", distancias_recomendadas)
```

```python
print("\nRanking de Servidores Recomendados:")

for posicao, indice in enumerate(indices_recomendados, start=1):
    servidor = catalogo_servidores[indice]
    nome = nomes_servidores[indice]
    distancia = distancias_recomendadas[posicao - 1]

    print(f"\n{posicao}º lugar: {nome}")
    print(f"vCPUs: {servidor[0]}")
    print(f"RAM: {servidor[1]} GB")
    print(f"SSD NVMe: {servidor[2]} GB")
    print(f"Distância: {distancia:.2f}")
```
